```python
from huggingface_hub import snapshot_download
import polars as pl
import glob
import tqdm as notebook_tqdm

local_dir = snapshot_download(
    repo_id="pipecat-ai/smart-turn-data-v3.2-train",
    repo_type="dataset",
)

parquet_paths = glob.glob(f"{local_dir}/**/*.parquet", recursive=True)
print(parquet_paths)

df = pl.read_parquet(parquet_paths)  # polars accepts a list of paths directly
print(df.shape)
```

```python
df.write_parquet("smart_turn_data_v3.2.parquet")
```
